In MC2 I liked how each notebook formed a complete pipeline: Go-board segmentation through K-means, morphology, and object measurement; torii-gate edge extraction with a hand-built Canny; and Japan-inspired augmentations using CLAHE, FFT high-boost, and Reinhard colour transfer.

What went especially well was iterating on preprocessing: distance-transform “circle” erosion gave clean Go stones and accurate counts, and CLAHE-then-sharpening made Shibuya both readable and stylised without feeling fake.

If I repeated the project, I’d automate parameter sweeps (for example K, blur size, thresholds) and push further into mapping Go stone centroids to grid coordinates and using the edge map for higher-level torii structure detection.

I was most impressed by how small morphological or frequency-domain changes dramatically affect downstream metrics and visual perception. I learned to think of vision systems as carefully tuned pipelines rather than single “magic” algorithms, and to justify each step in terms of the final application.
