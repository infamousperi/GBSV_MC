# Final Reflection MC2

## What went well:

- I built a full pipeline for the Go board, from K-means segmentation to morphology and object measurements.
- Distance-transform–based “circle” erosion cleaned up the stones, so counting and sizing became reliable.
- The augmentation notebook (CLAHE, FFT high-boost, Reinhard colour transfer) felt coherent and visually convincing.

## What I’d do differently:

- I’d script parameter sweeps for K, blur size, and thresholds instead of tuning them by hand.
- I’d push the Go-board work further by snapping stone centroids to grid intersections and exporting move lists.
- For the torii gate, I’d connect edges into higher-level shapes rather than stopping at the Canny stage.

## What impressed me:

- Small tweaks in morphology or frequency filters radically changed both recognition accuracy and perceived “style.”
- Seeing classic techniques like distance transforms and FFTs drive concrete tasks made them feel much less abstract.

## What I learned:

- Good vision systems are pipelines; every step must justify its place.
- Visual inspection is not enough link choices to measurements and downstream tasks.
