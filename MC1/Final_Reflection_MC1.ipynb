# Final Reflection MC1

## What went well:
I clearly showed why sampling above twice the max frequency avoids aliasing.

Turning the Nyquist–Shannon idea into pictures and simple examples helped it make sense.

I also fixed my mix-ups between correlation and convolution by focusing on the “flip” and what it changes in practice.

## What I’d do differently:
I’d start with tiny, simple signals before jumping to full images.

That would have saved time on debugging. I’d also add numbers (like PSNR/SSIM) to back up the visuals, not just rely on what looks right.

## What impressed me:
Small choices like zero-padding and how you treat borders can change results a lot.

Putting outputs side by side made those differences obvious.

## What I learned:
Always sample based on the task’s bandwidth.

Pick kernels that match the goal (denoise vs. sharpen). And be precise: am I correlating or convolving?

These habits will make my future image-processing work clearer, faster, and easier to reproduce.
